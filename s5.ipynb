5. Concatenate two DataFrames representing 'today_orders' and 'yesterday_orders' (each with columns: order_id, item, price), and display the combined DataFrame.<br><br><em><strong>Constraint:</strong> Use pd.concat() and reset the index after concatenation.</em>

```python
# Q1

import pandas as pd
from sqlalchemy import create_engine

engine = create_engine('sqlite:///zomato.db')

pd.DataFrame({
    'id': [1, 2, 3, 4, 5, 6],
    'name': ['Spice Hub', 'Pizza Point', 'Biryani House', 'Dosa Corner', 'Burger Barn', 'Cafe Mocha'],
    'city': ['Ahmedabad', 'Mumbai', 'Hyderabad', 'Chennai', 'Delhi', 'Pune'],
    'rating': [4.2, 3.9, 4.6, 4.1, 3.7, 4.4]
}).to_sql('restaurants', engine, if_exists='replace', index=False)

df = pd.read_sql('restaurants', engine)

print(df.head())
```

```python
# Q2

pd.DataFrame({
    'name': ['Movie A', 'Movie B', 'Movie C', 'Movie D'],
    'rating': [8.5, 7.2, 9.1, 6.8]
}).to_sql('movies', engine, if_exists='replace', index=False)

query = "SELECT name, rating FROM movies WHERE rating > 8"

df = pd.read_sql_query(query, engine)

print(df)
```

```python
# Q3

df = pd.read_json('https://jsonplaceholder.typicode.com/users')

print(df['username'])
```

```python
# Q4

from pathlib import Path

orders_path = Path('orders.csv')
users_path = Path('users.csv')

orders = pd.read_csv(orders_path)
users = pd.read_csv(users_path)

merged = pd.merge(orders, users, on='user_id')

print(merged[['username', 'amount']])
```

```python
# Q5

today_orders = pd.DataFrame({
    'order_id': [201, 202, 203],
    'item': ['Pizza', 'Burger', 'Biryani'],
    'price': [350, 180, 280]
})

yesterday_orders = pd.DataFrame({
    'order_id': [198, 199, 200],
    'item': ['Pasta', 'Dosa', 'Sandwich'],
    'price': [260, 120, 150]
})

combined = pd.concat([today_orders, yesterday_orders]).reset_index(drop=True)

print(combined)
```
