```python
# Q1

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

delivery_times = np.random.randint(15, 60, 50)

sns.histplot(delivery_times, bins=10)
plt.title('Zomato Delivery Times')
plt.xlabel('Delivery Time (minutes)')
plt.ylabel('Number of Orders')
plt.show()
```

```python
# Q2

tips = sns.load_dataset('tips')

sns.boxplot(x='day', y='total_bill', data=tips)
plt.title('Total Bill by Day of the Week')
plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.show()
```

```python
# Q3

sns.set_theme(style='darkgrid')

teams = ['CSK', 'MI', 'RCB', 'KKR', 'DC', 'RR', 'SRH', 'PBKS']
ipl = pd.DataFrame({
    'team': np.repeat(teams, 20),
    'runs': np.random.randint(100, 220, 160)
})

sns.violinplot(x='team', y='runs', data=ipl)
plt.title('IPL Run Distribution by Team')
plt.xlabel('Team')
plt.ylabel('Runs')
plt.show()
```

```python
# Q4

genres = ['Pop', 'Rock', 'Hip-Hop', 'Bollywood', 'Classical']
songs = pd.DataFrame({
    'genre': np.random.choice(genres, 40)
})

sns.countplot(x='genre', data=songs)
plt.title('Number of Songs per Genre')
plt.xlabel('Genre')
plt.ylabel('Count')
plt.show()
```

```python
# Q5

sns.set_theme(style='whitegrid')

steps = [6500, 8200, 7400, 9100, 10200, 5800, 8700]

sns.kdeplot(steps, color='green', fill=True)
plt.title('Distribution of Daily Step Counts')
plt.xlabel('Steps')
plt.ylabel('Density')
plt.show()
```
