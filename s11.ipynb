5. Add a custom text annotation to a Matplotlib chart showing Flipkart's monthly sales, marking the highest sales point with the label 'Big Billion Days'.<br><br><em><strong>Hint:</strong> Use the ax.annotate() function to place the label at the correct data point.</em>

```python
# Q1

import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 2, figsize=(10, 8))

axes[0, 0].plot([1, 2, 3, 4, 5], [10, 15, 13, 18, 20], marker='o')
axes[0, 0].set_title('Line Chart')

axes[0, 1].bar(['A', 'B', 'C', 'D'], [25, 40, 30, 55], color='orange')
axes[0, 1].set_title('Bar Chart')

axes[1, 0].scatter([1, 2, 3, 4, 5, 6], [5, 7, 4, 8, 6, 9], color='red')
axes[1, 0].set_title('Scatter Plot')

axes[1, 1].pie([30, 25, 20, 25], labels=['Swiggy', 'Zomato', "Domino's", 'Others'], autopct='%1.1f%%')
axes[1, 1].set_title('Pie Chart')

plt.tight_layout()
plt.show()
```

```python
# Q2

apps = ['Zomato', 'Swiggy', "Domino's"]
delivery_times = [32, 28, 25]
colors = ['red', 'orange', 'blue']
linestyles = ['-', '--', ':']
linewidths = [2, 3, 4]

fig, ax = plt.subplots()

for i in range(3):
    ax.bar(apps[i], delivery_times[i], color=colors[i], edgecolor='black',
           linestyle=linestyles[i], linewidth=linewidths[i])

ax.set_title('Average Delivery Time Comparison')
ax.set_xlabel('App')
ax.set_ylabel('Delivery Time (minutes)')
plt.show()
```

```python
# Q3

influencers = ['Infl A', 'Infl B', 'Infl C', 'Infl D', 'Infl E']
followers = [1200000, 850000, 2300000, 540000, 1750000]
daily_posts = [2.5, 1.8, 3.2, 1.2, 2.9]

fig, ax1 = plt.subplots()

ax1.plot(influencers, followers, color='blue', marker='o')
ax1.set_xlabel('Influencers')
ax1.set_ylabel('Followers', color='blue')

ax2 = ax1.twinx()
ax2.plot(influencers, daily_posts, color='red', marker='s')
ax2.set_ylabel('Average Daily Posts', color='red')

plt.title('Followers vs Daily Posts')
plt.show()
```

```python
# Q4

movies = ['Pathaan', 'Jawan', 'Animal', 'Dunki', 'Fighter']
tickets_sold = [5.5, 6.8, 5.2, 3.9, 2.8]
imdb = [5.9, 7.0, 6.2, 6.5, 6.4]

plt.scatter(tickets_sold, imdb, color='purple')

for i in range(len(movies)):
    plt.annotate(movies[i], (tickets_sold[i], imdb[i]), textcoords='offset points', xytext=(0, 8), ha='center')

plt.title('Tickets Sold vs IMDB Rating')
plt.xlabel('Tickets Sold (millions)')
plt.ylabel('IMDB Rating')
plt.show()
```

```python
# Q5
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
sales = [120, 110, 130, 125, 140, 135, 145, 150, 160, 310, 180, 170]

fig, ax = plt.subplots()
ax.plot(months, sales, marker='o', color='green')

max_index = sales.index(max(sales))
ax.annotate('Big Billion Days', xy=(months[max_index], sales[max_index]),
            xytext=(months[max_index - 3], sales[max_index] - 20),
            arrowprops=dict(facecolor='black', arrowstyle='->'))

ax.set_title("Flipkart's Monthly Sales")
ax.set_xlabel('Month')
ax.set_ylabel('Sales (in crores)')
plt.show()
```
