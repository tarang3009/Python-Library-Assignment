```python
# Q1

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

ipl = pd.DataFrame({
    'match_id': range(1, 11),
    'total_runs': [345, 312, 298, 367, 285, 330, 402, 275, 318, 356]
})

print(ipl['total_runs'].mean())
print(ipl['total_runs'].median())
print(ipl['total_runs'].min())
print(ipl['total_runs'].max())
print(ipl['total_runs'].std())
```

```python
# Q2

reviews = pd.DataFrame({
    'rating': np.random.choice([1, 2, 3, 4, 5], 100, p=[0.05, 0.10, 0.15, 0.30, 0.40]),
    'category': np.random.choice(['Electronics', 'Fashion', 'Home', 'Books'], 100)
})

sns.countplot(x='rating', data=reviews, order=[1, 2, 3, 4, 5])
plt.title('Number of Reviews by Rating')
plt.xlabel('Rating (Stars)')
plt.ylabel('Number of Reviews')
plt.show()
```

```python
# Q3

zomato = pd.DataFrame({
    'average_cost_for_two': [300, 500, 800, 1200, 400, 1500, 600, 900, 350, 1000, 700, 1800],
    'user_rating': [3.8, 4.0, 4.2, 4.5, 3.9, 4.6, 4.1, 4.3, 3.7, 4.4, 4.0, 4.7]
})

sns.scatterplot(x='average_cost_for_two', y='user_rating', data=zomato)
plt.title('Cost for Two vs User Rating')
plt.xlabel('Average Cost for Two (₹)')
plt.ylabel('User Rating')
plt.show()
```

```python
# Q4

movies = pd.DataFrame({
    'genre': ['Action', 'Comedy', 'Drama', 'Action', 'Thriller', 'Comedy', 'Drama', 'Thriller', 'Action', 'Comedy'],
    'box_office_collection': [450, 180, 220, 520, 310, 150, 260, 340, 480, 200]
})

avg_collection = movies.groupby('genre')['box_office_collection'].mean().sort_values(ascending=False).reset_index()

print(avg_collection)
```

```python
# Q5

spotify = pd.DataFrame({
    'danceability': np.random.uniform(0.3, 0.95, 100),
    'energy': np.random.uniform(0.2, 0.98, 100),
    'valence': np.random.uniform(0.1, 0.95, 100),
    'popularity': np.random.randint(20, 100, 100)
})

sns.pairplot(spotify)
plt.show()

# One interesting pattern is that danceability and valence tend to move together in real Spotify data,
# since happier songs are usually more danceable. In this random sample no clear pattern appears,
# because the values are generated randomly.
```
