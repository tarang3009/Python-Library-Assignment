```python
# Q1

import pandas as pd
from ydata_profiling import ProfileReport

df = pd.read_csv('spotify_top_100_songs.csv')

profile = ProfileReport(df, title='Spotify Top 100 Songs Report')
profile.to_file('spotify_report.html')

# The report shows that the dataset has no major missing values, and the 'release_date' column is stored as
# an object type instead of datetime, so it needs to be converted.
```

```python
# Q2

import sweetviz as sv

mumbai = pd.read_csv('zomato_mumbai.csv')
delhi = pd.read_csv('zomato_delhi.csv')

report = sv.compare([mumbai, 'Mumbai'], [delhi, 'Delhi'])
report.show_html('zomato_comparison.html')

# One key difference is that Mumbai restaurants have a higher average cost for two than Delhi restaurants,
# while Delhi has a larger number of restaurants in the lower price range.
```
