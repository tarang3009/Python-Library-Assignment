```python
# Q1

import numpy as np

friend1 = np.array([5000, 7200, 6500, 8000, 9100, 4300, 10000])
friend2 = np.array([6000, 6800, 7000, 7500, 8200, 5200, 9500])

print(friend1 + friend2)
print(friend1 - friend2)
print(friend1 * friend2)
print(friend1 / friend2)
```

```python
# Q2

user_pref = np.array([
    [4, 3, 5],
    [2, 5, 1],
    [3, 4, 2]
])

song_pop = np.array([
    [5, 2, 4],
    [3, 4, 1],
    [2, 5, 3]
])

dot_result = np.dot(user_pref, song_pop)
matmul_result = np.matmul(user_pref, song_pop)

print(dot_result)
print(matmul_result)

# For 2D matrices, dot() and matmul() give the same result.
# The difference appears with higher dimensional arrays: dot() sums over the last axis of the first
# array and the second-last axis of the second, while matmul() treats them as stacks of matrices
# and multiplies them pairwise. matmul() also does not allow scalar multiplication.
```

```python
# Q3

image = np.array([
    [120, 130, 140, 150],
    [100, 110, 125, 135],
    [90, 105, 115, 128],
    [80, 95, 108, 120]
])

rotated = image.T

print(rotated)
print(np.mean(rotated))
print(np.median(rotated))
print(np.std(rotated))
print(np.var(rotated))
```

```python
# Q4

rating_grid = np.array([
    [1.0, 0.8, 0.6],
    [0.8, 1.0, 0.7],
    [0.6, 0.7, 1.0]
])

inverse = np.linalg.inv(rating_grid)
determinant = np.linalg.det(rating_grid)
eigenvalues, eigenvectors = np.linalg.eig(rating_grid)

print(inverse)
print(determinant)
print(eigenvalues)
print(eigenvectors)
```

```python
# Q5

orders = np.array([
    [120, 135, 150, 160, 145, 170],
    [100, 110, 125, 140, 130, 155]
])

reshaped = orders.reshape(3, 4)
flattened = reshaped.flatten()
part1, part2 = np.split(flattened, 2)
stacked = np.vstack((part1, part2))

print(reshaped)
print(flattened)
print(part1, part2)
print(stacked)
```
