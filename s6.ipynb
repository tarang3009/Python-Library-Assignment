2. Using the loaded IPL player stats DataFrame, apply dropna(axis=0, how='any') to remove all rows with any missing data and display the shape of the DataFrame before and after dropping.
3. For the 'runs' column in your IPL player stats DataFrame, use fillna() to replace missing values with the mean of the column. Print the updated column to verify the changes.

5. Pick any one column with missing values from your dataset and explain which missingness mechanism (MCAR, MAR, or MNAR) is most likely and why, using 2-3 sentences.<br><br><em><strong>Hint:</strong> Think about whether the missing data is random or related to other factors in the dataset.</em>

```python
# Q1

import pandas as pd
import numpy as np

df = pd.DataFrame({
    'player_name': ['Virat Kohli', 'Rohit Sharma', 'MS Dhoni', 'KL Rahul', 'Hardik Pandya', 'Shubman Gill', 'Rishabh Pant', 'Jos Buttler'],
    'runs': [7263, 6211, np.nan, 4689, 2525, np.nan, 3284, 3223],
    'matches': [237, 243, 250, 132, 123, 91, np.nan, 96],
    'team': ['RCB', 'MI', 'CSK', 'LSG', 'GT', np.nan, 'DC', 'RR']
})

print(df.isnull().sum())
print(df.notnull().sum())
```

```python
# Q2

print(df.shape)

df_dropped = df.dropna(axis=0, how='any')

print(df_dropped.shape)
```

```python
# Q3

df['runs'] = df['runs'].fillna(df['runs'].mean())

print(df['runs'])
```

```python
# Q4

ratings = pd.DataFrame({
    'restaurant': ['Spice Hub', 'Pizza Point', 'Biryani House', 'Dosa Corner', 'Burger Barn', 'Cafe Mocha'],
    'rating': [np.nan, 4.2, np.nan, 3.8, np.nan, 4.5]
})

print(ratings)

ratings['rating'] = ratings['rating'].ffill()
ratings['rating'] = ratings['rating'].bfill()

print(ratings)
```
